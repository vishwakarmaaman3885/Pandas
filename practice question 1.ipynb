```python
# Q3. Create a Pandas DataFrame that contains the following data:
# Name Alice, Bob, Claire
# Age 25, 30, 27
# Gender Female, Male, Female
df = pd.DataFrame({
    "Name":["Alice","Bob","Claire"],
    "Age":[25,30,27],
    "Gender":["Female","Male","Female"]
})
print(df) 
```

```python
# Q4. What is ‘DataFrame’ in pandas and how is it different from pandas.series? Explain with an example.
# DataFrame: A DataFrame is a two-dimensional data structure in Pandas that stores data in rows and columns, like an Excel table.
# Series: A Series is a one-dimensional data structure that stores a single sequence of values with index labels.
import pandas as pd

# Series
s = pd.Series([25, 30, 27])
print("Series:")
print(s)

# DataFrame
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Claire"],
    "Age": [25, 30, 27]
})
print("\nDataFrame:")
print(df)
```

```python
# Create a DataFrame using multiple Series. Explain with an example.
names = pd.Series(["Aman","Ayush","Khushi"])
ages = pd.Series([23,45,67])
marks = pd.Series([45,67,89])

df=pd.DataFrame({
    "Name":names,
    "Age":ages,
    "Marks":marks
})
print(df) 
```

```python
# Which of the following is mutable in nature: Series, DataFrame, Panel?
# Answer: All three are mutable in nature in the traditional Pandas classification.
# - Series: One-dimensional labeled data structure.
# - DataFrame: Two-dimensional tabular data structure.
# - Panel: Three-dimensional data structure, which was deprecated in Pandas 0.25 and removed in Pandas 1.0.
```
