```python
# python pandas - window functions
```

```python
df4 = pd.DataFrame({'a':[1,2,3,4,5,6,7,8,9]}) 
```

```python
df4
```

```python
df4.rolling(window=1).mean() 
# rolling() is used to calculate a value over a sliding window of rows. It is commonly used in
# data analysis to calculate moving averages and smooth fluctuations in data.
```

```python
df4.rolling(window=2).mean() 
```

```python
df4.rolling(window=3).mean() 
```

```python
df4.rolling(window=3).sum() 
```

```python
df4.rolling(window=3).min() 
```

```python
df4.cumsum()
# cumsum() means cumulative sum. It adds each value to the sum of all previous values.
# In simple words, it keeps a running total.
```
